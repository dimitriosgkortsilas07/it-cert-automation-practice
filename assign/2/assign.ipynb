## Άσκηση 1
Γράψτε μια συνάρτηση με το όνομα `myfunction1`, η οποία υπολογίζει το κοινό σημείο τριών επιπέδων στο χώρο, όταν τα επίπεδα δίνονται με την μορφή:

$$
\begin{aligned}
E_x x&+&E_y y&+&E_z z&=&1\\
(a+1)x&+&(b+1)y&+&(c+1)z&=&1\\
(a-2)x&+&(b-2)y&+&(c+5)z&=&1\\
\end{aligned}
$$

όπου:

* $E_x$, $E_y$ και $E_z$, οι συντελεστές του πρώτου επιπέδου που δίνονται ως διάνυσμα ndarray στην συνάρτηση,
* a,b,c τα 3 τελευταία ψηφία του αριθμού μητρώου σας (c=τελευταίο).

Η συνάρτηση επιστρέφει ένα διάνυσμα ndarray $P=[x,y,z]$.

Ελέγξτε τα αποτελέσματα της συνάρτησης για:

- $E=[10,0,0]$
- $E=[0,10,0]$


````python
# Λύση της άσκησης 1
# - Σε αυτό το κελί γράψτε μόνο τις ζητούμενες συναρτήσεις χωρίς print και input.

def myfunction1(L: np.ndarray) -> np.ndarray:
    a=0
    b=0
    c=2
    A=np.array([[L[0],L[1],L[2]],
       [a+1,b+1,c+1],
       [a-2,b-2,c+5]])
    det_A=np.linalg.det(A)
    if det_A==0:
      return None
    else:
      B=np.array([[1],[1],[1]])
      result=np.linalg.solve(A,B)
      return (result.flatten().tolist())
````

````python
# Πρόχειρο άσκησης 1
# - Σε αυτό το κελί μπορείτε να τυπώσετε μεταβλητές και να κάνετε γραφικές παραστάσεις για επαλήθευση. Δεν λαμβάνεται υπόψη στην βαθμολόγηση.

print(myfunction1(np.array([10.,0.,0.])), myfunction1(np.array([0.,10.,0.])))
````

## Άσκηση 2
Γράψτε μια συνάρτηση `myfunction2` που επιλύει το γραμμικό σύστημα:

$$
\begin{bmatrix}
25 & -10 & 0 & -5 &\\
-10 & 20 & -10 & 0\\
0 & -10 & 20 & -10\\
-5 & 0 & -10 & 15
\end{bmatrix}
\cdot
\begin{bmatrix} x_1\\ x_2\\ x_3 \\ x_4 \end{bmatrix}
=
\begin{bmatrix} a+1\\ b+2\\ c+3 \\ 15 \end{bmatrix}
$$
όπου a,b,c τα 3 τελευταία ψηφία του αριθμού μητρώου σας (c=τελευταίο).

Η συνάρτηση θα δέχεται ένα διάνυσμα $x_{init}$ με τις αρχικές τιμές και θα επιστρέφει ένα διάνυσμα 4 στοιχείων $[x_1,x_2,x_3,x_4]$ με τα τελικά αποτελέσματα.
Χρησιμοποιήστε την μέθοδο Gauss-Seidel σταματώντας όταν όλα τα απόλυτα σφάλματα της τρέχουσας επανάληψης σε σχέση με την προηγούμενη είναι μικρότερα από 2e-3.
Αν το πλήθος των επαναλήψεων φτάσει τις 100 χωρίς να ικανοποιηθεί το κριτήριο σύγκλισης, η συνάρτηση θα πρέπει να επιστρέφει nan σε όλες τις τιμές του διανύσματος.

Ελέγξτε τα αποτελέσματα της συνάρτησης για $x_{init}=[0.,0.,0.,0.]$ και $x_{init}=[100.,100.,100.,100.]$.

```
Υπόδειξη: Χρησιμοποιήστε την μέθοδο .copy(), όταν θέλετε να αντιγράψετε τις τιμές σε διαφορετικό χώρο μνήμης και όχι να δημιουργήσετε ένα δείκτη σε υπάρχοντα πίνακα.
```
```
Ερώτηση για τις εξετάσεις: Ποιο διάνυσμα αρχικών τιμών δινει πιο γρήγορη σύγκλιση και γιατί;
```

````python
# Λύση της άσκησης 2
# - Σε αυτό το κελί γράψτε μόνο τις ζητούμενες συναρτήσεις χωρίς print και input.

def myfunction2(xinit: np.ndarray)->np.ndarray:
    a=0
    b=0
    c=2
    xnew=np.zeros(4)
    i=0
    xnew[0]=(a+1+10*xinit[1]+3*xinit[3])/25
    xnew[1]=(b+2+10*xnew[0]+3*xinit[2])/20
    xnew[2]=(c+3+10*xnew[1]+10*xinit[3])/20
    xnew[3]=(15+10*xnew[2]+5*xnew[0])/15
    while np.max(np.abs(xnew - xinit)) > 0.0001 and i<100:
        i+=1
        xinit=xnew.copy()
        xnew[0]=(a+1+10*xinit[1]+3*xinit[3])/25
        xnew[1]=(b+2+10*xnew[0]+3*xinit[2])/20
        xnew[2]=(c+3+10*xnew[1]+10*xinit[3])/20
        xnew[3]=(15+10*xnew[2]+5*xnew[0])/15
    if i>100:
      xnew = np.full(4,None,dtype=object)
    return xnew
````

````python
# Πρόχειρο άσκησης 2
# - Σε αυτό το κελί μπορείτε να τυπώσετε μεταβλητές και να κάνετε γραφικές παραστάσεις για επαλήθευση. Δεν λαμβάνεται υπόψη στην βαθμολόγηση.

print(myfunction2(np.array([0., 0., 0., 0.])),myfunction2(np.array([100., 100., 100., 100.])))
````
